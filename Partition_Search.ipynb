```python
#using pentagonal number theorem to find p(n) quickly

def partitions(N):
    p = [1] + [0] * N
    for n in range(1, N + 1):
        total = 0
        k = 1
        while True:
            g1 = k * (3 * k - 1) // 2
            if g1 > n:
                break
            g2 = k * (3 * k + 1) // 2
            sign = 1 if k % 2 == 1 else -1
            total += sign * p[n - g1]
            if g2 <= n:
                total += sign * p[n - g2]
            k += 1
        p[n] = total
    return p
```

```python
#checks if there is p(y) such that p(z) - p(x) = p(y)

def find_solutions(N):
    p = partitions(N)

    index_of = {p[n]: n for n in range(1, N + 1)}

    solutions = []
    for z in range(2, N + 1):
        for x in range(1, z):
            target = p[z] - p[x]
            if target < p[x]:
                break
            y = index_of.get(target)
            if y is not None:
                solutions.append((x, y, z))
    return solutions

start = time.time()
N = 50000

solutions = find_solutions(N)
print(f"Solutions with z <= {N}:")
for s in solutions:
    print(s)
```

```python

#we can find solutions faster by using upper and lower bounds to narrow the search p(x) <= p(z)/2 and p(x) < p(z) - p(z-1)

from bisect import bisect_left, bisect_right

def find_solutions_fast(N):
    p = partitions(N)
    index_of = {p[n]: n for n in range(1, N + 1)}

    solutions = []
    for z in range(2, N + 1):
        lo = bisect_left(p, p[z] - p[z - 1], 1, z)
        hi = bisect_right(p, p[z] // 2, 1, z)
        for x in range(lo, hi):
            y = index_of.get(p[z] - p[x])
            if y is not None:
                solutions.append((x, y, z))
    return solutions

N = 100000
sols = find_solutions_fast(N)
print(f"Solutions with z <= {N}:")
for s in sols:
    print(s)
```
